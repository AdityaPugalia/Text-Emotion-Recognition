```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

### Simple Bert Classifier Results

```python
#load results for simple bert
import pickle
with open('results/emotion_bert_results.pkl', 'rb') as f:
    train_accuracies, train_losses, val_accuracies, val_losses, test_accuracy, test_loss = pickle.load(f)
```

```python
import matplotlib.pyplot as plt
import numpy as np

early_stop_epoch = np.argmin(val_losses)

fig, axs = plt.subplots(1, 2, figsize=(12, 8))

# Accuracy Plot
axs[0].plot(train_accuracies, label='Train Accuracy')
axs[0].plot(val_accuracies, label='Validation Accuracy')
axs[0].axvline(x=early_stop_epoch, color='red', linestyle='--', label='Early Stopping')
axs[0].set_title('Train and Validation Accuracy')
axs[0].set_xlabel('Epoch')
axs[0].set_ylabel('Accuracy')
axs[0].legend()

# Loss Plot
axs[1].plot(train_losses, label='Train Loss')
axs[1].plot(val_losses, label='Validation Loss')
axs[1].axvline(x=early_stop_epoch, color='red', linestyle='--', label='Early Stopping')
axs[1].set_title('Train and Validation Loss')
axs[1].set_xlabel('Epoch')
axs[1].set_ylabel('Loss')
axs[1].legend()

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(4, 2))
ax.axis('off')  # hide axes

# Data for table
data = [
    ['Metric', 'Value'],
    ['Test Accuracy', f'{test_accuracy:.4f}'],
    ['Test Loss', f'{test_loss:.4f}']
]

# Create the table
table = ax.table(cellText=data, loc='center', cellLoc='center', colWidths=[0.4, 0.4])
table.auto_set_font_size(False)
table.set_fontsize(12)
table.scale(1.2, 1.5)  # adjust cell size

plt.title('Test Results', fontsize=14)
plt.tight_layout()
plt.show()

```
